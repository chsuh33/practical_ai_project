**인쇄본 90쪽 — 코드**

````python
from tensorflow.keras.datasets import cifar10
(X_train,y_train),(X_test,y_test)=cifar10.load_data()
print(X_train.shape)
print(X_test.shape)
print(y_train.shape)
print(y_test.shape)
````

**인쇄본 90쪽 — 출력 예시 (실행 코드 아님)**

```text
(50000, 32, 32, 3)
```

**인쇄본 91쪽 — 출력 예시 (실행 코드 아님)**

```text

(10000, 1)
```

**인쇄본 91쪽 — 코드**

````python
X_train = X_train / 255.0
X_test = X_test / 255.0
````

**인쇄본 92쪽 — 코드** —

````python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input
from tensorflow.keras.layers import Conv2D, MaxPool2D
from tensorflow.keras.layers import Flatten, Dense
model_alexnet = Sequential([
    Input(shape=(32, 32, 3)),
    Conv2D(
        48,
        kernel_size=(3, 3),
        strides=(1, 1),
        padding='same',
        activation='relu'
    ),
    MaxPool2D(
        pool_size=(2, 2),
        strides=(2, 2)
    ),
    Conv2D(
        96,
        kernel_size=(3, 3),
        strides=(1, 1),
        padding='same',
        activation='relu'
    ),
    MaxPool2D(
        pool_size=(2, 2),
        strides=(2, 2)
    ),
    Conv2D(
        192,
        kernel_size=(3, 3),
        padding='same',
        activation='relu'
    ),
    Conv2D(
        192,
        kernel_size=(3, 3),
        padding='same',
        activation='relu'
    ),
    Conv2D(
        256,
        kernel_size=(3, 3),
        padding='same',
        activation='relu'
    ),
    MaxPool2D(
        pool_size=(2, 2),
        strides=(2, 2)
    ),
    Flatten(),
    Dense(512, activation='relu'),
    Dense(256, activation='relu'),
    Dense(10, activation='softmax')
])
````

**인쇄본 94쪽 — 코드**

````python
from tensorflow.keras.optimizers import Adam
opt = Adam(
    learning_rate=0.001,
    beta_1=0.9,
    beta_2=0.999
)
model_alexnet.compile(
    optimizer=opt,
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
````

**인쇄본 94쪽 — 코드**

````python
model_alexnet.fit(
    X_train,
    y_train,
    epochs=20
)
````

**인쇄본 94쪽 — 코드**

````python
test_loss, test_accuracy = model_alexnet.evaluate(
    X_test,
    y_test
)
print(test_accuracy)
````

**인쇄본 95쪽 — 코드**

````python
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, Conv2D, MaxPool2D
from tensorflow.keras.layers import AveragePooling2D
from tensorflow.keras.layers import BatchNormalization, ReLU, Add
from tensorflow.keras.layers import Flatten, Dense
````

**인쇄본 95쪽 — 코드**

````python
inputs = Input(shape=(32, 32, 3))
x = Conv2D(
    32,
    kernel_size=(3, 3),
    strides=(1, 1),
    padding='same',
    use_bias=False
)(inputs)
````

**인쇄본 96쪽 — 코드**

````python
x = BatchNormalization()(x)
x = ReLU()(x)
x = Conv2D(
    32,
    kernel_size=(3, 3),
    strides=(1, 1),
    padding='same',
    use_bias=False
)(x)
x = BatchNormalization()(x)
x = ReLU()(x)
````

**인쇄본 96쪽 — 코드**

````python
x = MaxPool2D(
    pool_size=(2, 2),
    strides=(2, 2)
)(x)
x = Conv2D(
    64,
    kernel_size=(3, 3),
    padding='same',
    use_bias=False
)(x)
x = BatchNormalization()(x)
x = ReLU()(x)
````

**인쇄본 96쪽 — 코드**

````python
skip = x
````

**인쇄본 97쪽 — 코드**

````python
x = Conv2D(
    64,
    kernel_size=(3, 3),
    padding='same',
    use_bias=False
)(x)
x = BatchNormalization()(x)
x = ReLU()(x)
x = Conv2D(
    64,
    kernel_size=(3, 3),
    padding='same',
    use_bias=False
)(x)
x = BatchNormalization()(x)
````

**인쇄본 97쪽 — 코드**

````python
x = Add()([x, skip])
x = ReLU()(x)
````

**인쇄본 98쪽 — 코드**

````python
x = MaxPool2D(
    pool_size=(2, 2),
    strides=(2, 2)
)(x)
x = Conv2D(
    128,
    kernel_size=(3, 3),
    padding='same',
    use_bias=False
)(x)
x = BatchNormalization()(x)
x = ReLU()(x)
skip = x
x = Conv2D(
    128,
    kernel_size=(3, 3),
    padding='same',
    use_bias=False
)(x)
x = BatchNormalization()(x)
x = ReLU()(x)
x = Conv2D(
    128,
    kernel_size=(3, 3),
    padding='same',
    use_bias=False
)(x)
x = BatchNormalization()(x)
x = Add()([x, skip])
x = ReLU()(x)
````

**인쇄본 98쪽 — 코드** — **원문 연결/문법 확인 필요:** '(' was never closed

````python
x = AveragePooling2D(
    pool_size=(8, 8)
    )(x)
x = Flatten()(x)
outputs = Dense(10,activation='softmax')(x)
````

**인쇄본 99쪽 — 코드**

````python
model_resnet = Model(inputs=inputs,outputs=outputs)
````

**인쇄본 99쪽 — 코드**

````python
opt = Adam(
    learning_rate=0.001,
    beta_1=0.9,
    beta_2=0.999
)
model_resnet.compile(
    optimizer=opt,
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
````

**인쇄본 99쪽 — 코드**

````python
model_resnet.fit(X_train,y_train,epochs=20)
````

**인쇄본 100쪽 — 코드**

````python
test_loss, test_accuracy = model_resnet.evaluate(
    X_test,
    y_test
)
print(test_accuracy)
````
