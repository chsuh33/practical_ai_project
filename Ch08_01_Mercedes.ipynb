**인쇄본 202쪽 — 코드**

````python
import pandas as pd
data = pd.read_csv('train.csv')
print(data.shape)
````

**인쇄본 202쪽 — 출력 예시 (실행 코드 아님)**

```text
(4209, 378)
```

**인쇄본 202쪽 — 코드**

````python
cols = ['ID', 'y', 'X0', 'X1', 'X2', 'X3',
        'X10', 'X11', 'X12', 'X13']
print(data[cols].head())
````

**인쇄본 202쪽 — 출력 예시 (실행 코드 아님)**

```text
   ID       y  X0 X1  X2 X3  X10  X11  X12  X13

4  13   78.02  az  v   n  f    0    0    0    0
```

**인쇄본 203쪽 — 코드**

````python
cf = data.select_dtypes(include=['object']).columns
print(cf)
````

**인쇄본 203쪽 — 출력 예시 (실행 코드 아님)**

```text
Index(['X0', 'X1', 'X2', 'X3', 'X4', 'X5', 'X6', 'X8'],
      dtype='object')
```

**인쇄본 203쪽 — 코드**

````python
data = pd.get_dummies(data, columns=cf, dtype=float)
print(data.shape)
````

**인쇄본 203쪽 — 출력 예시 (실행 코드 아님)**

```text
(4209, 565)
```

**인쇄본 204쪽 — 코드**

````python
X_df = data.drop(['ID', 'y'], axis=1)
t_df = data['y']
print(X_df.shape)
print(t_df.shape)
````

**인쇄본 204쪽 — 출력 예시 (실행 코드 아님)**

```text
(4209, 563)
(4209,)
```

**인쇄본 204쪽 — 코드**

````python
tau = t_df.median()
y_df = (t_df <= tau).astype(int)
print(tau)
print(y_df.value_counts())
````

**인쇄본 204쪽 — 출력 예시 (실행 코드 아님)**

```text
99.15
0    2103
1    2106
Name: count, dtype: int64
```

**인쇄본 205쪽 — 코드**

````python
X = X_df.to_numpy(dtype=float)
y = y_df.to_numpy()
print(X.shape)
print(y.shape)
````

**인쇄본 205쪽 — 출력 예시 (실행 코드 아님)**

```text
(4209, 563)
(4209,)
```

**인쇄본 205쪽 — 코드**

````python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.1,
    stratify=y,
    random_state=0
)
print(X_train.shape)
print(X_test.shape)
print(y_train.shape)
print(y_test.shape)
````

**인쇄본 206쪽 — 출력 예시 (실행 코드 아님)**

```text
(3788, 563)
(421, 563)
(3788,)
(421,)
```

**인쇄본 207쪽 — 코드**

````python
from sklearn.linear_model import LogisticRegression
lr_clf = LogisticRegression(max_iter=5000)
lr_clf.fit(X_train, y_train)
test_performance = lr_clf.score(X_test, y_test)
print(test_performance)
````

**인쇄본 207쪽 — 출력 예시 (실행 코드 아님)**

```text
0.905
```

**인쇄본 208쪽 — 코드**

````python
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Input, Dense
model = Sequential([
    Input(shape=(563,)),
    Dense(250, activation='relu'),
    Dense(1, activation='sigmoid')
])
````

**인쇄본 208쪽 — 코드**

````python
model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)
model.fit(X_train,y_train,batch_size=128,epochs=30,verbose=0)
````

**인쇄본 208쪽 — 코드**

````python
test_loss, test_performance = model.evaluate(X_test,y_test,verbose=0)
print(test_performance)
````
