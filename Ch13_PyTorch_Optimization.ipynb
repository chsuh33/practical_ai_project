**인쇄본 313쪽 — 코드**

````python
import torch
x = torch.tensor([1.0], requires_grad=True)
alpha = 0.1
num_iter = 20
for i in range(num_iter):
    f = x**2
    f.backward()
    with torch.no_grad():
        x -= alpha * x.grad
        x.grad.zero_()
    if (i + 1) % 5 == 0:
        print(i + 1, x.item())
````

**인쇄본 314쪽 — 코드**

````python
import torch
x = torch.ones(2, requires_grad=True)
alpha = 0.1
num_iter = 20
for i in range(num_iter):
    f = torch.norm(x)**2
    f.backward()
    with torch.no_grad():
        x -= alpha * x.grad
        x.grad.zero_()
    if (i + 1) % 5 == 0:
        print(i + 1, x.detach().numpy())
````

**인쇄본 315쪽 — 코드**

````python
import torch
x1 = torch.tensor([1.0], requires_grad=True)
x2 = torch.tensor([1.0])
alpha = 0.1
num_iter = 20
for i in range(num_iter):
    x = torch.cat([x1, x2])
    f = torch.norm(x)**2
    f.backward()
    with torch.no_grad():
        x1 -= alpha * x1.grad
        x1.grad.zero_()
    if (i + 1) % 5 == 0:
        print(i + 1, x1.item(), x2.item())
````

**인쇄본 316쪽 — 코드**

````python
import torch
x = torch.zeros(2, requires_grad=True)
v = torch.zeros(1, requires_grad=True)
alpha = 0.1
beta = 0.2
k = 5

num_iter = 50
for i in range(num_iter):
    for _ in range(k):
        L = torch.norm(x)**2 + v * (1 - torch.sum(x))
        L.backward()
        with torch.no_grad():
            x -= alpha * x.grad
        x.grad.zero_()
        v.grad.zero_()
    L = torch.norm(x)**2 + v * (1 - torch.sum(x))
    L.backward()
    with torch.no_grad():
        v += beta * v.grad
    x.grad.zero_()
    v.grad.zero_()
    if (i + 1) % 10 == 0:
        print(i + 1, x.detach().numpy(), v.item())
        
````
