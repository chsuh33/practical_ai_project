**인쇄본 129쪽 — 코드**

````python
from sklearn.datasets import load_iris
iris = load_iris()
X = iris.data
y = iris.target
print(X.shape)
print(iris.feature_names)
````

**인쇄본 129쪽 — 출력 예시 (실행 코드 아님)**

```text

```

**인쇄본 129쪽 — 코드**

````python
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=0
)
tree_clf = DecisionTreeClassifier(max_depth=2)
tree_clf.fit(X_train, y_train)
````

**인쇄본 130쪽 — 코드**

````python
test_performance = tree_clf.score(X_test, y_test)
print(test_performance)
````

**인쇄본 130쪽 — 출력 예시 (실행 코드 아님)**


**인쇄본 130쪽 — 코드**

````python
from sklearn.tree import plot_tree
import matplotlib.pyplot as plt
plt.figure(figsize=(10, 6))
plot_tree(
    tree_clf,
    feature_names=iris.feature_names,
    class_names=iris.target_names,
    filled=True
)
plt.show()
````

**인쇄본 131쪽 — 코드**

````python
for name, score in zip(
    iris.feature_names,
    tree_clf.feature_importances_
):
    print(name, ':', score)
````

````python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(
    X_reg, y_reg, test_size=0.2, random_state=0
)
````

````python
from sklearn.tree import DecisionTreeRegressor
tree_reg = DecisionTreeRegressor(max_depth=4)
tree_reg.fit(X_train, y_train)
````

**인쇄본 137쪽 — 코드**

````python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
iris = load_iris()
X_iris = iris.data
````

**인쇄본 138쪽 — 코드**

````python
y_iris = iris.target
X_iris_train, X_iris_test, y_iris_train, y_iris_test = train_test_split(
    X_iris, y_iris, test_size=0.2, random_state=0
)
rnd_clf = RandomForestClassifier(
    n_estimators=500,
    max_leaf_nodes=16,
    bootstrap=False,
    random_state=0,
    n_jobs=-1
)
rnd_clf.fit(X_iris_train, y_iris_train)
test_performance = rnd_clf.score(X_iris_test, y_iris_test)
print(test_performance)
````


**인쇄본 139쪽 — 코드**

**인쇄본 139쪽 — 코드**

````python
from sklearn.datasets import fetch_california_housing
from sklearn.ensemble import RandomForestRegressor
cali_prices = fetch_california_housing()
X_reg = cali_prices.data
y_reg = cali_prices.target
X_train, X_test, y_train, y_test = train_test_split(
    X_reg, y_reg, test_size=0.01
)
````

**인쇄본 140쪽 — 코드**

````python
for_reg = RandomForestRegressor()
for_reg.fit(X_train, y_train)
````

**인쇄본 140쪽 — 코드**

````python
from sklearn.metrics import mean_squared_error
y_pred_train = for_reg.predict(X_train)
y_pred_test = for_reg.predict(X_test)
mse_train = mean_squared_error(y_train, y_pred_train)
mse_test = mean_squared_error(y_test, y_pred_test)
print(mse_train)
print(mse_test)
````

**인쇄본 140쪽 — 출력 예시 (실행 코드 아님)**

```text
0.03454954220428868
0.3383040535545396
```

**인쇄본 140쪽 — 코드**

````python
import numpy as np
rmse_test = np.sqrt(mse_test)
nrmse_test = rmse_test / np.std(y_test)
````

**인쇄본 141쪽 — 코드**

````python
print(rmse_test)
print(nrmse_test)
````

**인쇄본 141쪽 — 코드** 

````python
from sklearn.model_selection import GridSearchCV
param_grid = {
    'n_estimators': [10, 50, 100, 300],
    'max_features': [1, 2, 3, 4]
}
forest_clf = RandomForestClassifier(
    max_depth=2,
    bootstrap=False,
    random_state=0
)
grid_search = GridSearchCV(
    forest_clf,
    param_grid,
    cv=5,
    scoring='accuracy',
    n_jobs=-1
)
grid_search.fit(X_iris_train, y_iris_train)
````

**인쇄본 142쪽 — 코드**

````python
print(grid_search.best_params_)
print(grid_search.best_estimator_)
print(grid_search.best_score_)
````

**인쇄본 143쪽 — 코드** 

````python
from sklearn.model_selection import RandomizedSearchCV
param_distributions = {
    'n_estimators': range(10, 301),
    'max_features': range(1, 5)
}
forest_clf = RandomForestClassifier(
    max_depth=2,
    bootstrap=False,
    random_state=0
)
randomized_search = RandomizedSearchCV(
    forest_clf,
    param_distributions,
    cv=5,
    n_iter=20,
    scoring='accuracy',
    random_state=0,
    n_jobs=-1
)
randomized_search.fit(X_iris_train, y_iris_train)
````

**인쇄본 144쪽 — 코드**

````python
print(randomized_search.best_params_)
print(randomized_search.best_estimator_)
print(randomized_search.best_score_)
````
