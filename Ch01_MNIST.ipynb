**인쇄본 39쪽 — 코드**

````python
from tensorflow.keras.datasets import mnist
(X_train, y_train),(X_test, y_test) = mnist.load_data()
print(X_train.shape)
print(X_test.shape)
````

**인쇄본 39쪽 — 출력 예시 (실행 코드 아님)**

```text
(60000, 28, 28)
(10000, 28, 28)
```

**인쇄본 40쪽 — 코드**

````python
X_train = X_train / 255
X_test = X_test / 255
````

**인쇄본 40쪽 — 코드**

````python
from sklearn.linear_model import RidgeClassifier
X_train_LS = X_train.reshape(-1, 28*28)
X_test_LS = X_test.reshape(-1, 28*28)
model_LS = RidgeClassifier()
model_LS.fit(X_train_LS, y_train)
accuracy_LS = model_LS.score(X_test_LS, y_test)
print(accuracy_LS)
````

**인쇄본 40쪽 — 출력 예시 (실행 코드 아님)**

```text
0.8604
```

**인쇄본 41쪽 — 코드**

````python
from sklearn.linear_model import LogisticRegression
model_LR = LogisticRegression(max_iter=1000)
model_LR.fit(X_train_LS, y_train)
accuracy_LR = model_LR.score(X_test_LS, y_test)
print(accuracy_LR)
````

**인쇄본 41쪽 — 출력 예시 (실행 코드 아님)**

```text
0.9256
```

**인쇄본 42쪽 — 코드**

````python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Flatten, Dense
model_DNN = Sequential([
    Flatten(input_shape=(28, 28)),
    Dense(128, activation='relu'),
    Dense(10, activation='softmax')
])
model_DNN.summary()
````

**인쇄본 42쪽 — 코드**

````python
from tensorflow.keras.optimizers import Adam
model_DNN.compile(
    optimizer=Adam(),
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
````

**인쇄본 42쪽 — 코드**

````python
model_DNN.fit(X_train, y_train, epochs=20)
````

**인쇄본 42쪽 — 코드**

````python
model_DNN.evaluate(X_test, y_test)
````
